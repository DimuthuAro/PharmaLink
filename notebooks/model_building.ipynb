```python
# 3. Function to clean text
def clean_text(x):
    if pd.isna(x):
        return ""
    x = str(x)
    x = re.sub(r'\s+', ' ', x)      # remove \n, \t, extra spaces
    x = re.sub(r'<.*?>', '', x)     # remove HTML tags
    x = x.strip()
    return x
```

```python
# 4. Clean all object (text) columns in the copy
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].apply(clean_text)
```

```python
# Combine all relevant text columns into a single NLP feature
text_cols = [
    'ProductIntroduction',
    'ProductUses',
    'ProductBenefits',
    'SideEffect',
    'HowToUse',
    'HowWorks',
    'QuickTips',
    'SafetyAdvice'
]

drug_clean['combined_text'] = drug_clean[text_cols].fillna('').agg(' '.join, axis=1)

drug_clean[['Name', 'Contains', 'combined_text']].head()
```

```python
def clean_html(x):
    if pd.isna(x):
        return ""
    x = str(x)

    # Remove real html tags <...>
    x = re.sub(r"<.*?>", " ", x)

    # Remove pseudo-html tags like: p dir="ltr", style="...", span ...
    x = re.sub(r'\b(?:p|span|div|dir|style|class|align)\b[^>]*', ' ', x, flags=re.IGNORECASE)

    # Remove leftover escaped quotes \" \'
    x = x.replace('\\"', ' ').replace("\\'", ' ')

    # Remove any multiple spaces/newlines
    x = re.sub(r"\s+", " ", x)

    return x.strip()
```

```python
drug_clean['combined_text'] = drug_clean['combined_text'].apply(clean_html)
drug_clean[['Name', 'combined_text']].head()
```

```python
def get_risk_label(text: str) -> int:
    risk = 0  # default = no clear food/alcohol interaction

    # ---------- HIGH RISK PATTERNS (label 2) ----------
    high_risk_words = [
        'do not take with alcohol',
        'do not drink alcohol',
        'avoid alcohol completely',
        'avoid alcohol',
        'serious liver damage',
        'severe liver damage',
        'stomach bleeding',
        'serious bleeding',
        'life-threatening',
        'grapefruit juice',
        'do not take with food',
        'on an empty stomach only',
        'contraindicated with alcohol',
        'fatal',
    ]

    if any(kw in text for kw in high_risk_words):
        risk = 2

    # ---------- MODERATE RISK PATTERNS (label 1) ----------
    moderate_words = [
        'take with food',
        'take it with food',
        'take this medicine with food',
        'may cause stomach upset',
        'to reduce stomach upset',
        'limit alcohol',
        'avoid drinking too much alcohol',
        'may cause dizziness',
        'may cause drowsiness',
        'use caution while driving',
        'avoid grapefruit',
        'take after food',
        'take before food',
    ]

    if any(kw in text for kw in moderate_words):
        # upgrade risk only if not already 2
        risk = max(risk, 1)

    return risk

drug_clean['risk_label'] = safety_text.apply(get_risk_label)

drug_clean[['Name', 'Contains', 'risk_label']].head(20)
```

```python
import joblib
import scipy.sparse as sp
import numpy as np

# Load saved files
rf_loaded = joblib.load("../model/food_drug_risk_model.pkl")
tfidf_loaded = joblib.load("../model/tfidf_vectorizer.pkl")
encoders_loaded = joblib.load("../model/category_encoders.pkl")

print("Model, vectorizer, and encoders loaded successfully!")
```

```python
cat_cols = ['Chemical_Class', 'Habit_Forming', 
            'Therapeutic_Class', 'Action_Class']

def predict_drug_risk(index):
    # 1. extract text
    text = drug_clean.loc[index, 'combined_text']
    
    # 2. vectorize text (sparse matrix)
    text_vec = tfidf_loaded.transform([text])   # shape (1, n_features)
    
    # 3. get categorical features and ensure numeric dtype
    cat_vals = drug_clean.loc[index, cat_cols].astype(float).values.reshape(1, -1)
    
    # 4. convert categorical to sparse matrix
    cat_sparse = sp.csr_matrix(cat_vals)
    
    # 5. combine text + categorical into one sparse row
    X_input = sp.hstack([text_vec, cat_sparse]).tocsr()
    
    # 6. predict risk label
    pred = rf_loaded.predict(X_input)[0]
    return pred
```

```python
# define nutrient columns again (to avoid NameError)
nutrient_cols = [
    'energy', 'protein', 'fat', 'carbs', 'fiber',
    'calcium', 'iron', 'vitamin_c', 'folate', 'vitamin_a', 'vitamin_e',
    'vitamin_k_proxy'
]

for col in nutrient_cols:
    food_subset[col] = pd.to_numeric(food_subset[col], errors='coerce').fillna(0.0)

food_subset['is_alcohol'] = food_subset['is_alcohol'].astype(int)
food_subset['is_leafy_green'] = food_subset['is_leafy_green'].astype(int)


food_subset.head()
```

```python
# add alcohol flag (1 = alcohol, 0 = not alcohol)
alcohol_words = ['beer', 'liquor', 'toddy', 'arak', 'wine', 'spirits']

def is_alcohol(name: str) -> int:
    name = str(name).lower()
    return int(any(word in name for word in alcohol_words))

food_subset['is_alcohol'] = food_subset['Food'].apply(is_alcohol)

food_subset.head(10)
```

```python
import re

def check_food_drug_interaction(drug_index, food_row):
    # --- drug info ---
    drug_name = drug_clean.loc[drug_index, 'Name']
    drug_contains = str(drug_clean.loc[drug_index, 'Contains']).lower()
    drug_text = str(drug_clean.loc[drug_index, 'combined_text']).lower()
    
    # vitamin K proxy from food
    vitk = float(food_row.get('vitamin_k_proxy', 0.0))

    # base risk from your ML drug model
    base_risk = predict_drug_risk(drug_index)
    risk = base_risk
    
    # --- food info ---
    calcium = float(food_row['calcium'])
    iron = float(food_row['iron'])
    fat = float(food_row['fat'])
    fiber = float(food_row['fiber'])
    is_alcohol = int(food_row.get('is_alcohol', 0))
    food_name = food_row['Food']
    
    # 1) Alcohol + CNS/sedative/antipsychotic drugs -> high risk
    cns_keywords = ['antipsychotic', 'antidepressant', 'antihistamine', 'sedative', 'antiepileptic']
    if is_alcohol == 1 and any(kw in drug_text for kw in cns_keywords):
        risk = max(risk, 2)
    
    # 2) Antibiotics + high calcium foods (milk, cheese etc.) -> moderate/high risk
    abx_keywords = ['amoxycillin', 'ciprofloxacin', 'tetracycline', 'doxycycline']
    if any(kw in drug_contains for kw in abx_keywords) and calcium > 200:
        risk = max(risk, 1)
    
    # 3) Thyroid medicines (levothyroxine) + high iron -> high risk
    if 'levothyroxine' in drug_contains and iron > 5:
        risk = max(risk, 2)
    
    # 4) Very high fat foods + drugs where fat affects absorption
    if fat > 20 and 'take on an empty stomach' in drug_text:
        risk = max(risk, 1)
    
    # 5) High fiber + slow absorption warning
    if fiber > 5 and 'slow absorption' in drug_text:
        risk = max(risk, 1)
    
    # 6) Warfarin / anticoagulant + high vitamin K leafy greens -> high risk
    if ('warfarin' in drug_contains or 'anticoagulant' in drug_text) and vitk > 100:
        risk = max(risk, 2)

    return int(risk)


# pick first 3 drugs and first 10 foods to test
for i in range(3):
    print("DRUG:", drug_clean.loc[i, "Name"])
    for _, row in food_subset.head(10).iterrows():
        r = check_food_drug_interaction(i, row)
        print("   Food:", row['Food'], "-> risk:", r)
    print("-" * 40)

```

```python
risk_map = {
    0: "Safe – No major interaction identified.",
    1: "Moderate – Use with caution and follow advice.",
    2: "High Risk – Avoid this combination; consult a doctor or pharmacist."
}

cns_keywords = ['antipsychotic', 'antidepressant', 'antihistamine', 'sedative', 'antiepileptic']
abx_keywords = ['amoxycillin', 'ciprofloxacin', 'tetracycline', 'doxycycline']

def explain_food_drug_interaction(drug_index, food_name):
    # locate food row
    row = food_subset[food_subset['Food'] == food_name].iloc[0]
    
    # run existing risk logic
    risk = check_food_drug_interaction(drug_index, row)
    
    # gather reasons (for explanation text)
    reasons = []
    
    drug_text = str(drug_clean.loc[drug_index, 'combined_text']).lower()
    drug_contains = str(drug_clean.loc[drug_index, 'Contains']).lower()
    
    calcium = row['calcium']
    iron = row['iron']
    fat = row['fat']
    fiber = row['fiber']
    is_alcohol = row['is_alcohol']
    vitk = float(row.get('vitamin_k_proxy', 0.0))
    
    # alcohol + CNS
    if is_alcohol == 1 and any(kw in drug_text for kw in cns_keywords):
        reasons.append("Alcohol can increase drowsiness and CNS side effects of this medicine.")
    
    # antibiotics + calcium
    if any(kw in drug_contains for kw in abx_keywords) and calcium > 200:
        reasons.append("High calcium in this food can reduce absorption of this antibiotic.")
    
    # thyroid + iron
    if 'levothyroxine' in drug_contains and iron > 5:
        reasons.append("High iron can reduce the effectiveness of thyroid medication.")
    
    # high fat
    if fat > 20 and 'take on an empty stomach' in drug_text:
        reasons.append("High fat may slow down or change absorption of this medicine.")
    
    # high fibre
    if fiber > 5 and 'slow absorption' in drug_text:
        reasons.append("High fibre may reduce absorption of this medicine.")

    # 6) vitamin K proxy + warfarin / anticoagulant
    if ('warfarin' in drug_contains or 'anticoagulant' in drug_text) and vitk > 100:
        reasons.append("Leafy green foods rich in vitamin K can reduce the blood-thinning effect of anticoagulant medicines.")
    
    # build message
    base_msg = risk_map[risk]
    if reasons:
        explanation = base_msg + " " + " ".join(reasons)
    else:
        explanation = base_msg + " No specific nutrient-triggered interaction was detected."
    
    return risk, explanation
```

```python
def recommend_safe_foods(drug_index, top_n=10):
    rows = []
    for _, row in food_subset.iterrows():
        r = check_food_drug_interaction(drug_index, row)
        if r == 0:
            rows.append(row)
    safe_df = pd.DataFrame(rows)
    
    # show only a few useful columns
    cols_show = ['Food', 'energy', 'protein', 'fat', 'carbs', 'fiber']
    return safe_df[cols_show].head(top_n)
```
