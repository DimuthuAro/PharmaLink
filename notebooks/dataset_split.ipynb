```python
import random
import shutil
from pathlib import Path

random.seed(42)

# 🔴 SOURCE: where your images are (Downloads)
SRC = Path(r"C:\Users\User\Downloads\archive (4)\Drug Vision\Data Combined")

# 🟢 DEST: inside your PharmaLink project
DST = Path(r"C:\Users\User\OneDrive - Sri Lanka Institute of Information Technology\Desktop\Research\PharmaLink\dataset")

splits = {"train": 0.7, "val": 0.15, "test": 0.15}

# Create split folders
for sp in splits:
    (DST / sp).mkdir(parents=True, exist_ok=True)

for drug_dir in SRC.iterdir():
    if not drug_dir.is_dir():
        continue

    images = list(drug_dir.glob("*.jpg")) + list(drug_dir.glob("*.png")) + list(drug_dir.glob("*.jpeg"))
    if len(images) < 20:
        print(f"⚠️ Skipping {drug_dir.name} (too few images: {len(images)})")
        continue

    random.shuffle(images)
    n = len(images)
    n_train = int(n * splits["train"])
    n_val = int(n * splits["val"])

    split_map = {
        "train": images[:n_train],
        "val": images[n_train:n_train + n_val],
        "test": images[n_train + n_val:]
    }

    for sp, files in split_map.items():
        out_dir = DST / sp / drug_dir.name
        out_dir.mkdir(parents=True, exist_ok=True)

        for f in files:
            shutil.copy(f, out_dir / f.name)  # ✅ copy only (safe)

    print(f"✅ {drug_dir.name}: {n} images split")

print("\n🎉 Done! Check PharmaLink/dataset/")
```

```python
import os, json
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
from tqdm.auto import tqdm

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

DATA_DIR = "../dataset"   # change if your dataset folder path is different
TRAIN_DIR = os.path.join(DATA_DIR, "train")
VAL_DIR   = os.path.join(DATA_DIR, "val")
TEST_DIR  = os.path.join(DATA_DIR, "test")

BATCH_SIZE = 32
IMG_SIZE = 224
EPOCHS = 10
LR = 1e-3
```

```python
train_tfms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(p=0.3),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

val_tfms = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])
```

```python
train_ds = datasets.ImageFolder(TRAIN_DIR, transform=train_tfms)
val_ds   = datasets.ImageFolder(VAL_DIR, transform=val_tfms)
test_ds  = datasets.ImageFolder(TEST_DIR, transform=val_tfms)

class_names = train_ds.classes
num_classes = len(class_names)
print("Classes:", class_names)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
val_loader   = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader  = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
```

```python
with open("class_names.json", "w") as f:
    json.dump(class_names, f)
print("Saved class_names.json")
```

```python
model = models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.DEFAULT)

# Replace classifier head
in_features = model.classifier[1].in_features
model.classifier[1] = nn.Linear(in_features, num_classes)

model = model.to(device)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
```

```python
def accuracy_from_logits(logits, y):
    preds = logits.argmax(dim=1)
    return (preds == y).float().mean().item()

best_val_acc = 0.0

for epoch in range(EPOCHS):
    # ---- TRAIN ----
    model.train()
    train_loss = 0.0
    train_acc = 0.0

    for x, y in tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS} [train]"):
        x, y = x.to(device), y.to(device)

        optimizer.zero_grad()
        logits = model(x)
        loss = criterion(logits, y)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()
        train_acc += accuracy_from_logits(logits, y)

    train_loss /= len(train_loader)
    train_acc  /= len(train_loader)

    # ---- VAL ----
    model.eval()
    val_loss = 0.0
    val_acc = 0.0

    with torch.no_grad():
        for x, y in tqdm(val_loader, desc=f"Epoch {epoch+1}/{EPOCHS} [val]"):
            x, y = x.to(device), y.to(device)
            logits = model(x)
            loss = criterion(logits, y)

            val_loss += loss.item()
            val_acc  += accuracy_from_logits(logits, y)

    val_loss /= len(val_loader)
    val_acc  /= len(val_loader)

    print(f"\nEpoch {epoch+1}: "
          f"train_loss={train_loss:.4f}, train_acc={train_acc:.4f} | "
          f"val_loss={val_loss:.4f}, val_acc={val_acc:.4f}")

    # save best
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), "drug_classifier_best.pth")
        print(f"✅ Saved best model (val_acc={best_val_acc:.4f})")
```

```python
model.load_state_dict(torch.load("drug_classifier_best.pth", map_location=device))
model.eval()

test_acc = 0.0
with torch.no_grad():
    for x, y in test_loader:
        x, y = x.to(device), y.to(device)
        logits = model(x)
        test_acc += accuracy_from_logits(logits, y)

test_acc /= len(test_loader)
print("✅ Test accuracy:", round(test_acc, 4))
```

```python
from PIL import Image
import torch.nn.functional as F

def predict_topk(image_path, topk=3):
    img = Image.open(image_path).convert("RGB")
    x = val_tfms(img).unsqueeze(0).to(device)

    model.eval()
    with torch.no_grad():
        logits = model(x)
        probs = F.softmax(logits, dim=1).cpu().numpy()[0]

    top_idx = probs.argsort()[::-1][:topk]
    return [(class_names[i], float(probs[i])) for i in top_idx]

# example:
# predict_topk("sample.jpg", topk=3)
```
