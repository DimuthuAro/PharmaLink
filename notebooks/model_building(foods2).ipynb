```python
food = food.drop_duplicates().reset_index(drop=True)
```

```python
food = food.drop(columns=['Unnamed: 36'])
```

```python
food = food.rename(columns={'drfoodlistdesc': 'Food'})
```

```python
food = food.rename(columns={
    'Energy (kcal)': 'Calories',
    'Total fats (g)': 'Fat',
    'Total carbohydrates (g)': 'Carbs',
    'Protein (g)': 'Protein'
})
```

```python
food['Food'].str.contains('GARLIC', case=False, na=False).sum()
food['Food'].str.contains('GINGER', case=False, na=False).sum()
food['Food'].str.contains('GINKGO', case=False, na=False).sum()
```
